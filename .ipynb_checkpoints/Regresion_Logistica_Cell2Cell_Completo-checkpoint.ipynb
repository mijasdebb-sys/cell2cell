# Regresión Logística — Cell2Cell

**Objetivo:** predecir la variable `Churn` (abandono del cliente) utilizando Regresión Logística.

- `cell2celltrain.csv`: datos de entrenamiento
- `cell2cellholdout.csv`: datos de evaluación (holdout)
- TARGET: `Churn`
  - `No = 0`
  - `Yes = 1`

**Nota importante:** el archivo `cell2cellholdout.csv` **no trae valores reales de `Churn`** (la columna viene vacía). Por eso, el modelo se entrena y se evalúa usando una partición interna de `cell2celltrain.csv` (train/test split). El holdout se usa únicamente al final para generar las predicciones finales, tal como ocurriría en un caso real de scoring sobre clientes nuevos.

```python
# Cargamos librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, roc_auc_score
```

```python
# Cargamos datos
cell2cell_train = pd.read_csv(
    'cell2celltrain.csv',
    sep=',',
    encoding='iso-8859-1'
)

cell2cell_holdout = pd.read_csv(
    'cell2cellholdout.csv',
    sep=',',
    encoding='iso-8859-1'
)

print("Train:", cell2cell_train.shape)
print("Holdout:", cell2cell_holdout.shape)
```

```python
# Primeras filas
cell2cell_train.head()
```

```python
# Información general
cell2cell_train.info()
```

```python
# Distribución del TARGET
print(cell2cell_train['Churn'].value_counts())
print("\nPorcentaje:")
print(cell2cell_train['Churn'].value_counts(normalize=True) * 100)
```

```python
# Visualizamos Churn
sns.countplot(data=cell2cell_train, x='Churn')
plt.title('Distribución de Churn')
plt.xlabel('Churn')
plt.ylabel('Cantidad de clientes')
plt.show()
```

```python
# Convertimos el TARGET a 0 y 1
# No = 0, Yes = 1
cell2cell_train['Churn'] = cell2cell_train['Churn'].map({
    'No': 0,
    'Yes': 1
})

# El holdout no trae valores reales de Churn (columna vacía).
# La dejamos convertida igual por consistencia, pero no se podrá usar para evaluar.
cell2cell_holdout['Churn'] = cell2cell_holdout['Churn'].map({
    'No': 0,
    'Yes': 1
})

print(cell2cell_train['Churn'].value_counts())
print("\nValores no nulos de Churn en holdout:", cell2cell_holdout['Churn'].notnull().sum())
```

```python
# Revisamos valores faltantes
missing = cell2cell_train.isnull().sum()
missing[missing > 0].sort_values(ascending=False)
```

```python
# Porcentaje de valores faltantes
missing_percent = (
    cell2cell_train.isnull().mean() * 100
).sort_values(ascending=False)

missing_percent[missing_percent > 0]
```

## Preparación de variables

- `CustomerID` se excluye porque es un identificador y no una variable predictora.
- `ServiceArea` se excluye porque tiene **747 categorías distintas**: convertirla con One-Hot Encoding generaría cientos de columnas casi vacías, sin aportar poder predictivo real y aumentando mucho el riesgo de overfitting.

```python
# Separamos variables predictoras (X) y TARGET (y)
columnas_excluidas = ['Churn', 'CustomerID', 'ServiceArea']

X_train = cell2cell_train.drop(columns=columnas_excluidas)
y_train = cell2cell_train['Churn']

X_holdout = cell2cell_holdout.drop(columns=columnas_excluidas)
y_holdout = cell2cell_holdout['Churn']  # queda vacío (NaN), no se usa para evaluar

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_holdout:", X_holdout.shape)
```

```python
# Identificamos variables numéricas y cualitativas
variables_numericas = X_train.select_dtypes(
    include=np.number
).columns

variables_cualitativas = X_train.select_dtypes(
    exclude=np.number
).columns

print("Variables numéricas:")
print(variables_numericas.tolist())

print("\nVariables cualitativas:")
print(variables_cualitativas.tolist())
```

## Imputación de valores faltantes

`LogisticRegression` de scikit-learn no admite valores `NaN`, así que antes de codificar hay que rellenarlos:

- **Numéricas:** se imputan con la **mediana** calculada en `X_train` (para no filtrar información del holdout hacia el entrenamiento).
- **Cualitativas:** se imputan con la **moda** (valor más frecuente) calculada en `X_train`.

Los mismos valores (medianas/modas de train) se aplican también al holdout.

```python
# Imputamos numéricas con la mediana de X_train
medianas = X_train[variables_numericas].median()

X_train[variables_numericas] = X_train[variables_numericas].fillna(medianas)
X_holdout[variables_numericas] = X_holdout[variables_numericas].fillna(medianas)

# Imputamos cualitativas con la moda de X_train
modas = X_train[variables_cualitativas].mode().iloc[0]

X_train[variables_cualitativas] = X_train[variables_cualitativas].fillna(modas)
X_holdout[variables_cualitativas] = X_holdout[variables_cualitativas].fillna(modas)

print("Faltantes en X_train tras imputar:", X_train.isnull().sum().sum())
print("Faltantes en X_holdout tras imputar:", X_holdout.isnull().sum().sum())
```

```python
# Convertimos variables categóricas a variables numéricas (One-Hot Encoding)
X_train_encoded = pd.get_dummies(
    X_train,
    drop_first=True
)

X_holdout_encoded = pd.get_dummies(
    X_holdout,
    drop_first=True
)

# Alineamos las columnas de train y holdout
X_train_encoded, X_holdout_encoded = X_train_encoded.align(
    X_holdout_encoded,
    join='left',
    axis=1,
    fill_value=0
)

# Convertimos a float
X_train_encoded = X_train_encoded.astype(float)
X_holdout_encoded = X_holdout_encoded.astype(float)

print("X_train_encoded:", X_train_encoded.shape)
print("X_holdout_encoded:", X_holdout_encoded.shape)
```

```python
# Revisamos valores faltantes después de la codificación
print("Valores faltantes en train:",
      X_train_encoded.isnull().sum().sum())

print("Valores faltantes en holdout:",
      X_holdout_encoded.isnull().sum().sum())
```

## Modelo de Regresión Logística

Como el holdout no tiene valores reales de `Churn`, dividimos `cell2celltrain.csv` en un conjunto de **entrenamiento** y uno de **prueba interna** (80/20, estratificado) para poder medir el desempeño del modelo (accuracy, matriz de confusión, reporte de clasificación).

Al final, se entrena un modelo definitivo con **todo** `cell2celltrain.csv` y se usa para generar las predicciones sobre `cell2cellholdout.csv`.

```python
# Dividimos train en entrenamiento (80%) y prueba interna (20%)
X_tr, X_test, y_tr, y_test = train_test_split(
    X_train_encoded,
    y_train,
    test_size=0.2,
    random_state=42,
    stratify=y_train
)

print("X_tr:", X_tr.shape)
print("X_test:", X_test.shape)
```

```python
# Creamos y entrenamos el modelo
modelo = LogisticRegression(
    max_iter=1000
)

modelo.fit(
    X_tr,
    y_tr
)

print("Modelo entrenado correctamente.")
```

```python
# Realizamos predicciones sobre el conjunto de prueba interna
y_pred = modelo.predict(X_test)

print("Primeras predicciones:")
print(y_pred[:20])
```

```python
# Accuracy
accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Accuracy:", accuracy)
```

```python
# Reporte de clasificación
print(classification_report(
    y_test,
    y_pred
))
```

```python
# Matriz de confusión
cm = confusion_matrix(
    y_test,
    y_pred
)

sns.heatmap(
    cm,
    annot=True,
    fmt='d'
)

plt.xlabel('Predicción')
plt.ylabel('Valor real')
plt.title('Matriz de Confusión - Regresión Logística')
plt.show()
```

```python
# Coeficientes del modelo
coeficientes = pd.DataFrame({
    'Variable': X_tr.columns,
    'Coeficiente': modelo.coef_[0]
})

coeficientes['Valor_Absoluto'] = coeficientes['Coeficiente'].abs()

coeficientes.sort_values(
    'Valor_Absoluto',
    ascending=False
).head(15)
```

## Interpretación

- **TARGET:** `Churn`
- `0` representa que el cliente **no abandonó**.
- `1` representa que el cliente **abandonó**.
- El modelo aprende la relación entre las variables de los clientes y `Churn`.
- `Accuracy` indica la proporción de predicciones correctas sobre el conjunto de prueba interna.
- La matriz de confusión permite observar los aciertos y errores de clasificación.
- Los coeficientes muestran la dirección de la relación de cada variable con la predicción del modelo; su interpretación debe hacerse considerando cómo fueron codificadas las variables (un coeficiente positivo aumenta la probabilidad de `Churn = 1`, uno negativo la reduce).
- El dataset está desbalanceado (hay muchos más clientes que no abandonan que clientes que sí), por lo que conviene mirar también `precision`/`recall` de la clase `1`, no solo el accuracy general.

## Predicciones finales sobre el holdout

Entrenamos el modelo definitivo con **todo** `cell2celltrain.csv` y generamos las predicciones sobre `cell2cellholdout.csv`. Como el holdout no tiene `Churn` real, aquí no se puede calcular accuracy: el resultado es el archivo de predicciones final.

```python
# Entrenamos el modelo final con todo el train
modelo_final = LogisticRegression(max_iter=1000)
modelo_final.fit(X_train_encoded, y_train)

# Predicciones (clase) y probabilidad de Churn=1 sobre el holdout
pred_holdout = modelo_final.predict(X_holdout_encoded)
proba_holdout = modelo_final.predict_proba(X_holdout_encoded)[:, 1]

resultados_holdout = pd.DataFrame({
    'CustomerID': cell2cell_holdout['CustomerID'],
    'Churn_Predicho': pred_holdout,
    'Probabilidad_Churn': proba_holdout
})

resultados_holdout.head(10)
```

```python
# Guardamos las predicciones en un CSV
resultados_holdout.to_csv('predicciones_holdout.csv', index=False)
print("Archivo 'predicciones_holdout.csv' generado con", resultados_holdout.shape[0], "predicciones.")
```
