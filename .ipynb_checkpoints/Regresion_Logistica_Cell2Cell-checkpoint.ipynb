# Regresión Logística — Cell2Cell

**Objetivo:** predecir la variable `Churn` (abandono del cliente) utilizando Regresión Logística.

- `cell2celltrain.csv`: datos de entrenamiento
- `cell2cellholdout.csv`: datos de evaluación (holdout)
- TARGET: `Churn`
  - `No = 0`
  - `Yes = 1`


```python
# Cargamos librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import statsmodels.api as sm
import statsmodels.formula.api as smf

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
# Cargamos datos
cell2cell_train = pd.read_csv(
    'cell2celltrain.csv',
    sep=',',
    encoding='iso-8859-1'
)

cell2cell_holdout = pd.read_csv(
    'cell2cellholdout.csv',
    sep=',',
    encoding='iso-8859-1'
)

print("Train:", cell2cell_train.shape)
print("Holdout:", cell2cell_holdout.shape)
```

```python
# Primeras filas
cell2cell_train.head()
```

```python
# Información general
cell2cell_train.info()
```

```python
# Distribución del TARGET
print(cell2cell_train['Churn'].value_counts())
print("\nPorcentaje:")
print(cell2cell_train['Churn'].value_counts(normalize=True) * 100)
```

```python
# Visualizamos Churn
sns.countplot(data=cell2cell_train, x='Churn')
plt.title('Distribución de Churn')
plt.xlabel('Churn')
plt.ylabel('Cantidad de clientes')
plt.show()
```

```python
# Convertimos el TARGET a 0 y 1
# No = 0, Yes = 1
cell2cell_train['Churn'] = cell2cell_train['Churn'].map({
    'No': 0,
    'Yes': 1
})

cell2cell_holdout['Churn'] = cell2cell_holdout['Churn'].map({
    'No': 0,
    'Yes': 1
})

print(cell2cell_train['Churn'].value_counts())
```

```python
# Revisamos valores faltantes
missing = cell2cell_train.isnull().sum()
missing[missing > 0].sort_values(ascending=False)
```

```python
# Porcentaje de valores faltantes
missing_percent = (
    cell2cell_train.isnull().mean() * 100
).sort_values(ascending=False)

missing_percent[missing_percent > 0]
```

```python
# Separamos variables predictoras (X) y TARGET (y)
# CustomerID se excluye porque es un identificador y no una variable predictora.

X_train = cell2cell_train.drop(
    columns=['Churn', 'CustomerID']
)

y_train = cell2cell_train['Churn']

X_holdout = cell2cell_holdout.drop(
    columns=['Churn', 'CustomerID']
)

y_holdout = cell2cell_holdout['Churn']

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_holdout:", X_holdout.shape)
print("y_holdout:", y_holdout.shape)
```

```python
# Identificamos variables numéricas y cualitativas
variables_numericas = X_train.select_dtypes(
    include=np.number
).columns

variables_cualitativas = X_train.select_dtypes(
    exclude=np.number
).columns

print("Variables numéricas:")
print(variables_numericas.tolist())

print("\nVariables cualitativas:")
print(variables_cualitativas.tolist())
```

```python
# Convertimos variables categóricas a variables numéricas (One-Hot Encoding)
X_train_encoded = pd.get_dummies(
    X_train,
    drop_first=True
)

X_holdout_encoded = pd.get_dummies(
    X_holdout,
    drop_first=True
)

# Alineamos las columnas de train y holdout
X_train_encoded, X_holdout_encoded = X_train_encoded.align(
    X_holdout_encoded,
    join='left',
    axis=1,
    fill_value=0
)

# Convertimos a float
X_train_encoded = X_train_encoded.astype(float)
X_holdout_encoded = X_holdout_encoded.astype(float)

print("X_train_encoded:", X_train_encoded.shape)
print("X_holdout_encoded:", X_holdout_encoded.shape)
```

```python
# Revisamos valores faltantes después de la codificación
print("Valores faltantes en train:",
      X_train_encoded.isnull().sum().sum())

print("Valores faltantes en holdout:",
      X_holdout_encoded.isnull().sum().sum())
```

## Modelo de Regresión Logística

Entrenamos el modelo con `cell2celltrain.csv` y posteriormente evaluamos sus predicciones usando `cell2cellholdout.csv`.


```python
# Creamos y entrenamos el modelo
modelo = LogisticRegression(
    max_iter=1000
)

modelo.fit(
    X_train_encoded,
    y_train
)

print("Modelo entrenado correctamente.")
```

```python
# Realizamos predicciones sobre holdout
y_pred = modelo.predict(X_holdout_encoded)

print("Primeras predicciones:")
print(y_pred[:20])
```

```python
# Accuracy
accuracy = accuracy_score(
    y_holdout,
    y_pred
)

print("Accuracy:", accuracy)
```

```python
# Reporte de clasificación
print(classification_report(
    y_holdout,
    y_pred
))
```

```python
# Matriz de confusión
cm = confusion_matrix(
    y_holdout,
    y_pred
)

sns.heatmap(
    cm,
    annot=True,
    fmt='d'
)

plt.xlabel('Predicción')
plt.ylabel('Valor real')
plt.title('Matriz de Confusión - Regresión Logística')
plt.show()
```

```python
# Coeficientes del modelo
coeficientes = pd.DataFrame({
    'Variable': X_train_encoded.columns,
    'Coeficiente': modelo.coef_[0]
})

coeficientes['Valor_Absoluto'] = coeficientes['Coeficiente'].abs()

coeficientes.sort_values(
    'Valor_Absoluto',
    ascending=False
).head(15)
```

## Interpretación

- **TARGET:** `Churn`
- `0` representa que el cliente **no abandonó**.
- `1` representa que el cliente **abandonó**.
- El modelo aprende la relación entre las variables de los clientes y `Churn`.
- `Accuracy` indica la proporción de predicciones correctas sobre el conjunto holdout.
- La matriz de confusión permite observar los aciertos y errores de clasificación.
- Los coeficientes muestran la dirección de la relación de cada variable con la predicción del modelo; su interpretación debe hacerse considerando cómo fueron codificadas las variables.
