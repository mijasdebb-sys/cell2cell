## 1. LIBRERÍAS Y FUNCIONES

## 2. PARÁMETROS

```python
RUTA_TRAIN   = 'cell2celltrain.csv'
RUTA_HOLDOUT = 'cell2cellholdout.csv'
TARGET       = 'Churn'
SEMILLA      = 42

# Variables numéricas que vamos a mirar con lupa en el análisis univariado
CLAVES = ['MonthlyRevenue', 'MonthlyMinutes', 'MonthsInService', 'CurrentEquipmentDays']
```

```python
tabla_faltantes = faltantes.copy()
tabla_faltantes['Porcentaje'].plot(kind='barh', color='#e74c3c')
plt.title('Porcentaje de valores faltantes por variable')
plt.xlabel('% de filas')
plt.gca().invert_yaxis()
plt.show()
```

```python
print('Filas duplicadas exactas :', df.duplicated().sum())
print('CustomerID duplicados    :', df['CustomerID'].duplicated().sum())
print('Duplicados sin CustomerID:', df.drop(columns='CustomerID').duplicated().sum())
```

```python
print('--- Inconsistencias detectadas ---')
print('MonthlyRevenue <= 0              :', (df['MonthlyRevenue'] <= 0).sum())
print('CurrentEquipmentDays < 0         :', (df['CurrentEquipmentDays'] < 0).sum(),
      '(valores entre -5 y -1)')
print('HandsetPrice == "Unknown"        :', (df['HandsetPrice'] == 'Unknown').sum(),
      f"({(df['HandsetPrice'] == 'Unknown').mean() * 100:.1f}%)")
print('PercChangeMinutes fuera de ±1000 :', (~df['PercChangeMinutes'].between(-1000, 1000)).sum())

print('\nPrimeros registros con CurrentEquipmentDays negativo:')
print(df.loc[df['CurrentEquipmentDays'] < 0,
             ['CustomerID', 'CurrentEquipmentDays']].head(10).to_string(index=False))
```

**Lectura de calidad:**

- Hay **3.515 celdas faltantes (0,12% del total)** repartidas en 14 variables. El bloque más importante es `AgeHH1`/`AgeHH2` con **909 casos (1,8%)**; le siguen `PercChangeMinutes`/`PercChangeRevenues` con 367 y seis variables de facturación con 156. Ninguna supera el 2%, así que **imputar es más razonable que eliminar filas**.
- **No hay filas duplicadas** ni `CustomerID` repetidos: cada fila es un cliente distinto.
- Hay **9 registros con `MonthlyRevenue` negativo** y **76 registros con `CurrentEquipmentDays` negativo** (entre -5 y -1). Son inconsistencias de captura, no valores de negocio: se documentan y se tratan como 0 días en los análisis por tramos.
- **896 registros tienen `PercChangeMinutes` fuera de ±1000%** (el rango va de -3.875 a 5.192). Son cambios de consumo extremos: conviene acotarlos, no borrarlos.
- `HandsetPrice` viene como texto y **el 56,8% es `"Unknown"`**: como variable es casi inútil tal como está. Se puede usar el resto como categórica, pero conviene no apoyarse en ella.
- `ServiceArea` con 747 niveles es el caso típico de variable de alta cardinalidad: mejor excluirla del one-hot.

**Decisión de preparación:** imputar numéricas con la **mediana** y categóricas con la **moda**, y descartar `CustomerID` del modelado.


### 4.3 Análisis univariado

¿Cómo se distribuye cada variable por separado? Buscamos sesgos, colas largas y valores atípicos.


```python
df[CLAVES + ['OverageMinutes', 'DroppedCalls']].describe().round(2)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 7))

for ax, col in zip(axes.ravel(), CLAVES):
    sns.histplot(df[col].dropna(), bins=40, kde=True, ax=ax, color='#3498db')
    ax.set_title(f'Distribución de {col}')
    ax.set_xlabel(col)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

for ax, col in zip(axes, ['MonthlyRevenue', 'MonthlyMinutes']):
    sns.boxplot(x=df[col].dropna(), ax=ax, color='#2ecc71')
    ax.set_title(f'Boxplot de {col} (los puntos son outliers)')

plt.tight_layout()
plt.show()
```

```python
# Conteo de outliers con la regla del rango intercuartílico (IQR)
resumen_outliers = []

for col in CLAVES + ['OverageMinutes', 'DroppedCalls']:
    s = df[col].dropna()
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    limite_sup = q3 + 1.5 * iqr
    n_out = ((s < q1 - 1.5 * iqr) | (s > limite_sup)).sum()
    resumen_outliers.append({
        'Variable': col,
        'Q1': round(q1, 2),
        'Q3': round(q3, 2),
        'Límite superior IQR': round(limite_sup, 2),
        'Outliers': n_out,
        '% del total': round(n_out / len(s) * 100, 1),
        'Máximo': round(s.max(), 2)
    })

pd.DataFrame(resumen_outliers)
```

**Lectura univariada:**

- `MonthlyRevenue`, `MonthlyMinutes` y `OverageMinutes` tienen **fuerte sesgo a la derecha**: la mayoría de los clientes paga poco y consume poco, pero existe una cola de clientes de altísimo consumo (hasta 1.223 de ingreso y 7.359 minutos).
- `MonthsInService` es más simétrica: los clientes tienen entre 6 y 61 meses, con mediana de 16.
- **Los outliers son reales, no errores**: un cliente que habla 7.000 minutos existe. En este dominio no se eliminan; se tratan con modelos robustos o transformación logarítmica cuando molesta.
- `OverageMinutes` tiene **11,5% de outliers**, pero porque su mediana es 3 minutos y el 25% de los clientes no paga excedentes. Es una distribución de "casi ceros".

**Consecuencia para el modelado:** las escalas son muy distintas (minutos en cientos, antigüedad en decenas). Cualquier modelo basado en distancias —K-Means incluido— **necesita escalado**.


### 4.4 El target: `Churn`

Antes de relacionar variables, hay que entender la variable que queremos predecir.


```python
conteo = df[TARGET].value_counts().to_frame('Clientes')
conteo['Porcentaje'] = (conteo['Clientes'] / len(df) * 100).round(2)
print(conteo)

sns.countplot(data=df, x=TARGET, hue=TARGET, palette=['#2ecc71', '#e74c3c'], legend=False)
plt.title('Distribución de Churn')
plt.ylabel('Cantidad de clientes')
plt.show()
```

**Lectura:** **71,18% no abandona y 28,82% sí**. Es un desbalance moderado.

¿Por qué importa? Porque un modelo que diga siempre "no se va" acierta el 71% y **no sirve para nada**: no detecta a un solo cliente en riesgo. Por eso los modelos de este proyecto usan `class_weight='balanced'` y se evalúan con **ROC-AUC, recall y precisión**, no solo con accuracy.


```python
def tasa_churn(col, top=10):
    t = (df.groupby(col)[TARGET]
           .agg(Clientes='size', Tasa_Churn=lambda s: (s == 'Yes').mean() * 100)
           .sort_values('Tasa_Churn', ascending=False)
           .round(1))
    return t.head(top)

for col in ['MadeCallToRetentionTeam', 'CreditRating', 'PrizmCode', 'Occupation']:
    print(f'\n--- {col} ---')
    print(tasa_churn(col).to_string())
```

**Lectura del target por segmento:**

- `MadeCallToRetentionTeam = Yes` → **45,0% de churn** contra 28,2% del resto. Es la señal categórica más fuerte del dataset, y tiene lógica: **llamar al equipo de retención es síntoma de que el cliente ya quería irse**. Ojo con la interpretación: es predictiva, no causal.
- `CreditRating` se comporta al revés de lo intuitivo: los clientes con peor rating (`5-Low`) tienen **22,1%** de churn, menos que los de mejor rating (`3-Good`, 31,0%). El crédito no es un buen discriminador de abandono en esta base.
- `PrizmCode`: Rural 31,7% vs Suburban 28,1%. Diferencia real pero chica.
- `Occupation`: Homemaker 32,5% vs Retired 25,2%, con tamaños muestrales muy desparejos (157 vs 733). **Cuidado con conclusiones sobre categorías pequeñas.**


### 4.5 Análisis bivariado

¿Qué variables se mueven junto con el churn?


```python
df['Churn_num'] = df[TARGET].map({'No': 0, 'Yes': 1})

correlaciones = (df[variables_numericas + ['Churn_num']]
                 .corr()['Churn_num']
                 .drop('Churn_num')
                 .sort_values())

top_corr = pd.concat([correlaciones.head(6), correlaciones.tail(6)]).sort_values()

plt.figure(figsize=(9, 6))
plt.barh(top_corr.index, top_corr.values,
         color=['#e74c3c' if v > 0 else '#3498db' for v in top_corr.values])
plt.axvline(0, color='black', linewidth=0.8)
plt.title('Variables numéricas más asociadas a Churn (correlación de Pearson)')
plt.xlabel('Correlación con Churn')
plt.tight_layout()
plt.show()

top_corr.round(3)
```

**Advertencia metodológica:** una correlación de 0,10 es **débil**. En este dataset ninguna variable numérica sola explica el churn: el fenómeno es multivariado. Que todas las correlaciones sean bajas no significa que no sirvan; significa que hay que combinarlas en un modelo.


```python
vars_heatmap = ['Churn_num', 'CurrentEquipmentDays', 'MonthsInService', 'MonthlyMinutes',
                'MonthlyRevenue', 'CustomerCareCalls', 'RetentionCalls', 'DroppedCalls',
                'OverageMinutes', 'PercChangeMinutes']

plt.figure(figsize=(10, 8))
sns.heatmap(df[vars_heatmap].corr(), annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=0.5)
plt.title('Matriz de correlaciones — variables clave')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, col in zip(axes, ['MonthlyMinutes', 'MonthsInService', 'CurrentEquipmentDays']):
    sns.boxplot(data=df, x=TARGET, y=col, hue=TARGET,
                palette=['#2ecc71', '#e74c3c'], legend=False, showfliers=False, ax=ax)
    ax.set_title(f'{col} según Churn')

plt.tight_layout()
plt.show()
```

```python
# Relación no lineal: tasa de churn por tramos
df['tramo_equipo'] = pd.cut(df['CurrentEquipmentDays'].clip(lower=0),
                            [-1, 90, 180, 365, 730, 3000],
                            labels=['0-90', '91-180', '181-365', '366-730', '>730'])
df['tramo_antiguedad'] = pd.cut(df['MonthsInService'], [0, 6, 12, 24, 36, 72],
                                labels=['0-6', '7-12', '13-24', '25-36', '37+'])

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

for ax, col in zip(axes, ['tramo_equipo', 'tramo_antiguedad']):
    tasa = df.groupby(col, observed=True)['Churn_num'].mean().mul(100)
    tasa.plot(kind='bar', ax=ax, color='#e67e22', edgecolor='white')
    ax.set_title(f'Tasa de churn (%) por {col}')
    ax.set_ylabel('% de churn')
    ax.tick_params(axis='x', rotation=0)
    for i, v in enumerate(tasa.values):
        ax.text(i, v + 0.5, f'{v:.1f}%', ha='center', fontsize=9)

plt.tight_layout()
plt.show()

print('Churn por tramo de equipo:')
print(df.groupby('tramo_equipo', observed=True)['Churn_num']
        .agg(Clientes='size', Tasa=lambda s: round(s.mean() * 100, 1)).to_string())
```

**Lectura bivariada — el hallazgo más sólido del EDA:**

`CurrentEquipmentDays` es la variable numérica más asociada al churn (r = 0,104) y la relación es **monótona**:

| Antigüedad del equipo | Tasa de churn |
|---|---:|
| 0-90 días | 21,2% |
| 91-180 días | 24,8% |
| 181-365 días | 26,6% |
| 366-730 días | 32,5% |
| más de 730 días | **36,1%** |

Es decir: **cuanto más viejo el teléfono del cliente, más probable que se vaya**. Tiene sentido de negocio: un equipo viejo es el momento natural para comparar ofertas de la competencia.

En cambio, la antigüedad en el servicio (`MonthsInService`) **no es lineal**: el churn sube hasta el tramo de 13-24 meses (30,8%) y después baja. Un modelo lineal no capturaría ese patrón; un árbol sí.


### 4.6 Generación de hipótesis y hallazgos

**Cinco hallazgos con evidencia:**

1. **El equipo viejo es el mejor predictor individual.** Churn 36,1% con equipos de más de 2 años vs 21,2% con equipos nuevos, y es la correlación numérica más alta (0,104).
2. **Llamar al equipo de retención es señal de fuga.** 45,0% de churn en quienes llamaron, contra 28,2% en el resto.
3. **El problema está desbalanceado** (71/29). La métrica de negocio no puede ser accuracy.
4. **Los outliers son legítimos** (clientes de alto consumo), no errores: se conservan.
5. **Hay variables que no aportan** (`CustomerID` es identificador; `ServiceArea` con 747 niveles es ruido; `HandsetPrice` tiene 56,8% "Unknown").

**Tres hipótesis para el modelado (formato del curso):**

> *"Creemos que* **la antigüedad del equipo** *influye en* **el abandono** *porque observamos que* la tasa de churn pasa de 21,2% a 36,1% según la antigüedad del teléfono."

> *"Creemos que* **haber contactado al equipo de retención** *influye en* **el abandono** *porque observamos que* ese grupo tiene 45,0% de churn frente a 28,2%."

> *"Creemos que* **el nivel de uso (minutos y excedentes)** *influye en* **el abandono** *porque observamos que* los clientes que se van tienen menor consumo mediano."

**Decisiones que salen del EDA:**

| Decisión | Justificación |
|---|---|
| Excluir `CustomerID` | Es un identificador, no una característica |
| Excluir `ServiceArea` del one-hot | 747 niveles, alta cardinalidad |
| Imputar mediana / moda | Faltantes < 2% por variable |
| Usar `class_weight='balanced'` | 71/29 de desbalance |
| Evaluar con ROC-AUC y recall | Accuracy es engañosa con desbalance |
| Escalar para K-Means | Escalas muy distintas entre variables |
